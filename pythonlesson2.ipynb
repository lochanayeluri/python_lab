```python
L1 = [1,2,3,4,'y',"lochana"]
L2 = [5,6,7,8,'p',"python"]
print(L1[4])          #indexing
print(L2[::-1])       #slicing
print(L1+L2)          #concetenation
print(L1*2)          #repetation
print(5 in L2)        #membership
print(5 in L1)
print(5 not in L1)
for i in L2:          #iteration
  print(i)
```
