```python
#program that prints reverse of the number using functions
def reverse(num):
  rev=0
  while (num!=0):
    rem=num%10
    rev=rev*10+rem
    num=num//10
  return rev
print(reverse(12345))
```

```python
#program that count the no of vowels and consonants in a string
```

```python
a=str(input())
b=a.lower()
vowels="aieou"
count_v=0
count_c=0
for ch in b:
  if ch.isalpha():
    if ch in vowels:
      count_v+=1
    else:
      count_c+=1
print("no of vowels",count_v)
print("no of consonants",count_c)
```
