```python
#Append
M1 = [6,2,0,'list']
M2=[1,9,3,'python']
M1.append("8")
print(M1)
```

```python
#Insert
M1.insert(2,1)
print(M1)
```

```python
#pop
M1.pop(2)
print(M1)
```

```python
# remove
M2.remove(3)
print(M2)
```

```python
# extend
M1.extend(M2)
print(M1)
```

```python
#join two lists
M3=M1+M2
print(M3)
```

```python
#copy lists
M2=M1.copy()
print(M2)
```

```python
M2[1:3]=('new','gen')
print(M2)
```

```python
#sort
S1=[3,1,9,6,4,9]
S1.sort()
print(S1)
```

```python
# count
S1.count(9)
```

```python
# index
S1.index(1)
```
