```python
#program that count the no of vowels & consonants in a string
word=str(input())
b=word.lower
vowel_count=0
consonant_count=0
vowels="aeiou"
for ch in word:
  if  ch.isalpha():
    if ch in vowels:
      vowel_count+=1
    else :
      consonant_count+=1
print("no of vowels: ",vowel_count)
print("no of consonants: ",consonant_count)
```
