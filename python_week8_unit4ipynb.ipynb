```python
#Write a program to sort words in a file and put them in another file. The output
#file should have only lower-case words, so any upper-case words from source
#must be lowered.
```

```python
with open("f1.txt","r") as f:
  data=f.read().split()
  print(data)
  data_lower=[i.lower() for i in data]
  sort_data=sorted(data_lower)
with open("copy.txt","w") as f2:
  for i in sort_data:
    f2.write(i+"\n")
print("done")
```
